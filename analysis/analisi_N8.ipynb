### Filtra i risultati per selezionare le istanze su cui eseguire le policies

```python
import pandas as pd
```

```python
dftot = pd.read_excel('Instances_list.xlsx')
print(len(dftot))
print(dftot.columns)
```

```python
len(dftot[dftot['solved']==False])
```

```python
solved_good = len(dftot[(dftot['gap']<=0.001) &(dftot['gap']>=0.0)])
solved_good_good_VSS = len(dftot[(dftot['gap']<=0.001) & (dftot['gap']>=0.0) & (dftot['VSS']>=4)])
print(f"solved with gap < 0.1%: {solved_good}")
print(f"solved with gap < 0.1% and with VSS >= 4%: {solved_good_good_VSS}")
```

```python
print(f"gap max: {dftot['gap'].max()}; gap average: {dftot[dftot['gap']>=0]['gap'].mean()}, time average: {dftot['runTime'].mean()}")
```

```python
ntot = len(dftot)
df = dftot[(dftot['solved']==True) & (dftot['gap']<=0.1)]
nsolved = len(df)
perc = 100*(nsolved/ntot)
vss_mean = df['VSS'].mean()
print(f"istanze totali: {ntot}; solved: {nsolved}; % solve: {perc:.2f}; average VSS: {vss_mean:.2f}")
```

```python
df['TMAX'] = df['name'].str[9:12]
df['CAP'] = df['name'].str[14:17]
df
```

```python
#rimuovi istanze con tmax = T15
dft = df# df[df['TMAX']!='T15']
dft
```

```python
mean_vss = dft['VSS'].mean()
ni = len(dft['VSS'])
print(f"Number of instances: {ni}: average vss: {mean_vss:.2f}")
```

```python
# instanze con VSS buono
threshold = 4
dfok = dft[dft['VSS']>=threshold]
mean_vss = dfok['VSS'].mean()
ni = len(dfok['VSS'])
print(f"Number of instances with VSS >= {threshold}: {ni}: average vss: {mean_vss:.2f}")
dfok
```

```python
dfok.to_excel('instances_list_N8_filtered.xlsx', index=None)
```

```python
# toglendo cap=150
dfc = dft[dft['CAP']!='150']
mean_vss = dfc['VSS'].mean()
ni = len(dfc['VSS'])
print(f"Number of instances: {ni}: average vss: {mean_vss:.2f}")
dfc
```
