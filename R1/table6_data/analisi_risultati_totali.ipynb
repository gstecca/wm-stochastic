### Aggrega e tabella i risultati dello stocastico e del'EVV per tutte le istanze in Instances_list.xlsx

```python
import pandas as pd
```

```python
dftot = pd.read_excel('Instances_list.xlsx')
print(len(dftot))
print(dftot.columns)
dftot
```

```python
# ['I3', 'N5', 'M3', 'T15', 'C150', 'DepCentral', 's0']
#N
dftot['N'] = dftot['name'].str.split('_').str[1]
dftot['N'] = dftot['N'].str[1:].astype(int) 
#M
dftot['M'] = dftot['name'].str.split('_').str[2]
dftot['M'] = dftot['M'].str[1:].astype(int)
#T
dftot['T'] = dftot['name'].str.split('_').str[3]
dftot['T'] = dftot['T'].str[1:].astype(int)
#C
dftot['C'] = dftot['name'].str.split('_').str[4]
dftot['C'] = dftot['C'].str[1:].astype(int)
#Depot
dftot['Depot'] = dftot['name'].str.split('_').str[5]
#seed
dftot['seed'] = dftot['name'].str.split('_').str[6]
dftot['seed'] = dftot['seed'].str[1:].astype(int)
dftot
```

```python
# Tabella con risultati di tutte e 720 le istanze, raggruppate per n. clienti,
# mostrando in particolare VSS min, media, max. numero risolte,
#numero non risolte, gap medio, minimo e max di quelle risolte, tempi dello
#stocastico, tempi dell’ EVV (deterministico)
#groupBY = ['N']
#groupBY = ['C']
groupBY = ['N', 'C']
dftot_n = dftot.groupby(groupBY).agg(
    num_instances=('name', 'count'),
    num_solved=('solved', 'sum'),
    VSS_min=('VSS', lambda x: round(x[x > 0].min(), 2)),
    VSS_max=('VSS', lambda x: round(x.max(), 2)),
    VSS_mean=('VSS', lambda x: round(x[x > 0].mean(), 2)),
    runTime_mean=('runTime', lambda x: round(x[x > 0].mean(), 2)),
    runTime_max=('runTime', lambda x: round(x.max(), 2)),
    gap_mean=('gap', lambda x: round(x[x > 0].mean(), 2)),
    gap_max=('gap', lambda x: round(x[x > 0].max(), 2)),
    runTime_mean_EEV=('runTimeEEV', lambda x: round(x[x > 0].mean(), 2)),
    runTime_max_EEV=('runTimeEEV', lambda x: round(x.max(), 2)),
    gap_mean_EEV=('gapEEV', lambda x: round(x[x > 0].mean(), 2)),
    gap_max_EEV=('gapEEV', lambda x: round(x[x > 0].max(), 2))
)
dftot_n['num_unsolved'] = dftot_n['num_instances'] - dftot_n['num_solved']
if len(groupBY) == 2:
    dftot_n.insert(0,'C','')
    dftot_n.insert(0,'N','')
    dftot_n['N'] = dftot_n.index.get_level_values(0)
    dftot_n['C'] = dftot_n.index.get_level_values(1)
    dftot_n.to_excel(f'Instances_list_agg_{"".join(groupBY)}.xlsx', index=False)
elif len(groupBY) == 1:
    dftot_n.to_excel(f'Instances_list_agg_{"".join(groupBY)}.xlsx', index=True)
dftot_n
```
