```python
import pandas as pd
```

```python
dftot = pd.read_csv('log_table_policies_p.csv')
print(len(dftot))
print(dftot.columns)
dftot
```

```python
"""
# ['I3', 'N5', 'M3', 'T15', 'C150', 'DepCentral', 's0']
#N
dftot['N'] = dftot['name'].str.split('_').str[1]
dftot['N'] = dftot['N'].str[1:].astype(int) 
#M
dftot['M'] = dftot['name'].str.split('_').str[2]
dftot['M'] = dftot['M'].str[1:].astype(int)
#T
dftot['T'] = dftot['name'].str.split('_').str[3]
dftot['T'] = dftot['T'].str[1:].astype(int)
#C
dftot['C'] = dftot['name'].str.split('_').str[4]
dftot['C'] = dftot['C'].str[1:].astype(int)
#Depot
dftot['Depot'] = dftot['name'].str.split('_').str[5]
#seed
dftot['seed'] = dftot['name'].str.split('_').str[6]
dftot['seed'] = dftot['seed'].str[1:].astype(int)
dftot
"""
```

```python
# Tabella con risultati di tutte e 2136 istanze per policy, mostrando tempo minimi, medio e massimo di esecuzione, gap e VSS
groupBY = ['type']
dftot_n = dftot.groupby(groupBY).agg(
    num_instances=('name', 'count'),
    mintime=('runtime_p', lambda x: round(x.min(), 2)),
    avgtime=('runtime_p', lambda x: round(x.mean(), 2)),
    maxtime=('runtime_p', lambda x: round(x.max(), 2))
)
#add a row with type ALL and average all values
dftot_n.loc['ALL'] = [
    dftot_n['num_instances'].sum(),
    round(dftot_n['mintime'].mean(), 2),
    round(dftot_n['avgtime'].mean(), 2),
    round(dftot_n['maxtime'].mean(), 2)
]

dftot_n.to_excel(f'Instances_list_agg_policies_runtime{"".join(groupBY)}2026.xlsx', index=True)
dftot_n
```
