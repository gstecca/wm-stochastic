```python
import pandas as pd
```

```python
dftot = pd.read_excel('Instances_list_N7.xlsx')
```

```python
len(dftot[dftot['gap']>0.001])
```

```python
dftot['gap'].max()
```

```python
ntot = len(dftot)
df = dftot[(dftot['solved']==True) & (dftot['gap']<=0.001)]
nsolved = len(df)
perc = 100*(nsolved/ntot)
vss_mean = df['VSS'].mean()
print(f"istanze totali: {ntot}; solved: {nsolved}; % solve: {perc:.2f}; average VSS: {vss_mean:.2f}")
```

```python
df['TMAX'] = df['name'].str[9:12]
df['CAP'] = df['name'].str[14:17]
df
```

```python
#rimuovi istanze con tmax = T15
dft = df[df['TMAX']!='T15']
dft
```

```python
mean_vss = dft['VSS'].mean()
ni = len(dft['VSS'])
print(f"Number of instances: {ni}: average vss: {mean_vss:.2f}")
```

```python
# instanze con VSS buono
# Lanciare le policies su queste
dfok = dft[dft['VSS']>=4]
mean_vss = dfok['VSS'].mean()
ni = len(dfok['VSS'])
print(f"Number of instances: {ni}: average vss: {mean_vss:.2f}")
dfok
```

```python
# toglendo cap=150
dfc = dft[dft['CAP']!='150']
mean_vss = dfc['VSS'].mean()
ni = len(dfc['VSS'])
print(f"Number of instances: {ni}: average vss: {mean_vss:.2f}")
dfc
```

```python
# instanze con VSS buono togliendo il cap150
dfokcap = dfc[dfc['VSS']>=5]
mean_vss = dfokcap['VSS'].mean()
ni = len(dfokcap['VSS'])
print(f"Number of instances: {ni}: average vss: {mean_vss:.2f}")
dfok
```
